## The situation

Bed 14 cost the hospital its trust in a screen. The response was not to fix AROGYA-1. The
response was paper.

Since February every drug given in the four wards has been written twice. Once by the model,
into the electronic record. Once by hand, in Kannada numerals, on a chart clipped to the foot
of the bed. Two independent witnesses. That was the whole point.

The charts were scanned in April and the originals were pulped, because the scan was the
record now and nobody wanted two.

**In June the archive drive failed.**

What came back off the platter is one tensor. Five hundred and twelve crops, twenty-eight by
twenty-eight, one Kannada digit each, still in ledger order. The scanner's calibration log did
not survive. The clean scans did not survive. The paper is pulp.

Sixty-four beds. Eight digits a bed.

One of those digits is wrong. The ward has been dosing off it since February, and the
electronic record - the witness that was supposed to disagree - agrees, because the
transcription error happened on the way *in*, before either witness saw it.

Restore the archive. Read the ledger. Find the digit.

---

## What you are given

**The archive.** It restores itself in the setup cell below.

| array | shape | what it is |
|---|---|---|
| `Z` | `(512, 28, 28)` | the recovered crops, in ledger order, float |
| `calib_noisy` | `(24, 28, 28)` | the scanner's calibration strip as it came off the platter |
| `calib_clean` | `(24, 28, 28)` | the same strip, clean. The only paired data you have |

plus the formulary, the check weights and the ward codes.

**The corpus.** The official Kannada-MNIST release. You fetch it yourself in the next cell. It
is the only clean Kannada handwriting you are permitted to learn from and it is the reason the
fetch is not optional.

That is the entire evidence base. There is no validation set, no labels for `Z`, and no clean
copy of anything in the archive.

---
## Step 0 - fetch the official corpus

Kannada-MNIST, Prabhu 2019. Primary source is the Kaggle competition **`Kannada-MNIST`**; the
cell falls back to the author's upstream release if Kaggle credentials are not configured.
Both routes are hash-verified against the same tensors, so it does not matter which one you
get - but it does matter that the hashes print `OK`. If they do not, stop and fix it before
you model anything.

On Colab, `kagglehub` will prompt you for your Kaggle username and key. If you would rather not
bother, let it fall through; the mirror needs no credentials.

````python
# ====================== STEP 0 - OFFICIAL CORPUS - run first ======================
import os, io, hashlib, urllib.request
import numpy as np

WANT = {                                     # sha256 of the raw array bytes
    "X_train": "ec528f157e171d9b", "y_train": "6222ecafc85bee75",
    "X_dig":   "2ce1659f85abab67", "y_dig":   "470b073476546add",
}
MIRROR = ("https://raw.githubusercontent.com/vinayprabhu/Kannada_MNIST/"
          "master/data/output_tensors/MNIST_format")
FILES = {"X_train": "X_kannada_MNIST_train", "y_train": "y_kannada_MNIST_train",
         "X_dig":   "X_dig_MNIST",           "y_dig":   "y_dig_MNIST"}


def _h(a):
    return hashlib.sha256(np.ascontiguousarray(a).tobytes()).hexdigest()[:16]


def _from_kaggle():
    import kagglehub                                          # route 1: official Kaggle
    p = kagglehub.competition_download("Kannada-MNIST")
    rd = lambda f: np.loadtxt(os.path.join(p, f), delimiter=",", skiprows=1, dtype=np.int64)
    tr, dg = rd("train.csv"), rd("Dig-MNIST.csv")
    return {"X_train": tr[:, 1:].reshape(-1, 28, 28).astype(np.uint8),
            "y_train": tr[:, 0].astype(np.uint8),
            "X_dig":   dg[:, 1:].reshape(-1, 28, 28).astype(np.uint8),
            "y_dig":   dg[:, 0].astype(np.uint8)}


def _from_mirror():
    out = {}
    for k, stem in FILES.items():                             # route 2: upstream release
        fn = stem + ".npz"
        if not os.path.exists(fn):
            urllib.request.urlretrieve(f"{MIRROR}/{fn}", fn)
        out[k] = np.load(fn)["arr_0"]
    return out


try:
    CORPUS = _from_kaggle(); route = "kaggle competition Kannada-MNIST"
except Exception as e:
    print("kaggle route unavailable (%s); using upstream mirror" % type(e).__name__)
    CORPUS = _from_mirror(); route = "upstream release"

print("source:", route)
ok = True
for k in FILES:
    g = _h(CORPUS[k]); good = (g == WANT[k]); ok &= good
    print(f"  {k:<8} {str(CORPUS[k].shape):<16} {g}  {'OK' if good else 'MISMATCH'}")
assert ok, "corpus integrity check failed - do not proceed"

X_TRAIN = CORPUS["X_train"].astype(np.float64) / 255.0        # 60000 clean, labelled
Y_TRAIN = CORPUS["y_train"].astype(np.int64)
X_DIG   = CORPUS["X_dig"].astype(np.float64) / 255.0          # 10240 relief-staff hand
Y_DIG   = CORPUS["y_dig"].astype(np.int64)
print("\ncorpus ready:", X_TRAIN.shape, X_DIG.shape)
````

## The chain that produced the archive

You get the chain. You do not get the parameters. Every constant below is drawn independently
per crop, from ranges you are not told, and none of them are recorded anywhere.

For one crop, source image `s`:

```
x0 = clip( warp(s, theta, scale, dx, dy), 0, 1 )                      <-- the ground truth
u  = g * hblur(x0, L) + b                                             optics, then sensor
z  = sqrt(abar_t) * u  +  sqrt(1 - abar_t) * eps ,  eps ~ N(0, I)     read noise
z  = saltpepper(z, p)                                                 bit rot
z  = clip(z, -0.35, 1.35)                                             stored range
```

`abar_t` is the cosine schedule with `T = 200`. `warp` and `hblur` are given below, exactly as
they were run. `theta`, `scale`, `dx`, `dy`, `L`, `g`, `b`, `t`, `p` are not.

Three things about this chain deserve to be said out loud.

**The target is `x0`, not `s`.** `x0` is a warped crop. It does not exist in Kannada-MNIST, it
does not exist in Dig-MNIST, and it does not exist in any file you can download. There is no
lookup. If you want it you have to produce it.

**The chain is not invertible.** The clip destroyed information and no estimator gets it back.
There is a floor. It is not zero. Nobody is going to reach it.

**`t` is different for every crop.** One denoiser tuned to one noise level will do well on part
of the archive and actively damage the rest. Restoration here is a *blind* problem: you have to
work out how badly each crop was hit before you can decide how hard to push on it.

---

## The ledger

The archive is in ledger order. Crop `8r + c` is digit `c` of row `r`.

```
row r        d0    ward code     1 CARD   2 GENM   3 ORTH   4 PEDS
             d1 d2 bed number    == r + 1, always, no exceptions
             d3 d4 drug code     must appear in the formulary
             d5 d6 dose in mg    must lie inside that drug's range
             d7    check digit
```

The check digit closes the row:

```
d7 == ( 7*d0 + 3*d1 + 1*d2 + 7*d3 + 3*d4 + 1*d5 + 7*d6 )  mod 10
```

**Exactly one row fails its check.** In that row exactly one of `d0` .. `d6` was mis-transcribed
on the way into the chart. The other sixty-three rows close.

The bed anchor is a gift and you should treat it as one: two of the eight digits in every row
are known before you read anything, which makes the archive its own labelled calibration set,
128 crops deep.

````python
# ========================= GIVEN - run as-is, do not edit =========================
import numpy as np

WEIGHTS = np.array([7, 3, 1, 7, 3, 1, 7])
WARDS   = [1, 2, 3, 4]                              # CARD GENM ORTH PEDS
FORMULARY = {                                       # drug code -> (dose_lo, dose_hi) mg
    "05": (10, 40), "08": (5, 25),  "11": (20, 60), "17": (15, 45),
    "23": (30, 90), "26": (10, 50), "34": (25, 75), "39": (5, 35),
    "42": (40, 95), "51": (20, 70), "63": (15, 55), "77": (35, 85),
}
T_STEPS = 200


def cosine_alpha_bar(t, T=T_STEPS, s=0.008):
    f = lambda u: np.cos(((u / T) + s) / (1 + s) * np.pi / 2) ** 2
    return float(np.clip(f(t) / f(0), 1e-6, 1.0))


def warp(img, theta_deg, scale, dx, dy):
    """Bilinear inverse-mapped rotate + scale + translate about the centre."""
    h, w = img.shape
    cy, cx = (h - 1) / 2.0, (w - 1) / 2.0
    th = np.deg2rad(theta_deg)
    ct, st = np.cos(th) / scale, np.sin(th) / scale
    yy, xx = np.meshgrid(np.arange(h), np.arange(w), indexing="ij")
    ry, rx = yy - cy - dy, xx - cx - dx
    sx = ct * rx + st * ry + cx
    sy = -st * rx + ct * ry + cy
    x0 = np.floor(sx).astype(int); y0 = np.floor(sy).astype(int)
    fx, fy = sx - x0, sy - y0
    out = np.zeros_like(img, dtype=np.float64)
    for oy in (0, 1):
        for ox in (0, 1):
            yi, xi = y0 + oy, x0 + ox
            ok = (yi >= 0) & (yi < h) & (xi >= 0) & (xi < w)
            wgt = (fy if oy else 1 - fy) * (fx if ox else 1 - fx)
            out[ok] += wgt[ok] * img[np.clip(yi, 0, h - 1), np.clip(xi, 0, w - 1)][ok]
    return out


def hblur(img, L):
    """Horizontal box streak of odd length L. L = 1 is a no-op."""
    if L <= 1:
        return img.copy()
    p = L // 2
    pad = np.pad(img, ((0, 0), (p, p)), mode="edge")
    return sum(pad[:, k:k + img.shape[1]] for k in range(L)) / L


def check_digit(body):
    """body = [d0..d6] -> the digit d7 must equal."""
    return int(np.dot(WEIGHTS, np.asarray(body)) % 10)


def row_is_legal(r, row8):
    """Structural validity of a full 8-digit row at ledger index r."""
    d = list(map(int, row8))
    if d[0] not in WARDS:                       return False
    if d[1] * 10 + d[2] != r + 1:               return False
    code = f"{d[3]}{d[4]}"
    if code not in FORMULARY:                   return False
    lo, hi = FORMULARY[code]
    return lo <= d[5] * 10 + d[6] <= hi


print("schedule:", " ".join("t=%d:%.3f" % (t, cosine_alpha_bar(t))
                            for t in (0, 50, 100, 150, 200)))
print("check_digit([1, 0, 1, 0, 5, 2, 5]) =", check_digit([1, 0, 1, 0, 5, 2, 5]))
print("row_is_legal(0, [1, 0, 1, 0, 5, 2, 5, 6]) =",
      row_is_legal(0, [1, 0, 1, 0, 5, 2, 5, 6]))
````

# The two tracks

Both are scored. They are not independent, and the coupling is deliberate: a restoration that
looks pleasant but destroys the stroke topology will read badly, and a reader that cannot
handle what your denoiser leaves behind will cost you on both.

---

### Track A - Restoration

Recover `x0` for all 512 crops.

Produce: `restored`, shape `(512, 28, 28)`, float, every value in `[0, 1]`.

Values outside `[0, 1]` are clipped before scoring rather than rejected, so nothing you submit
will error - it will just quietly score worse than you expected.

---

### Track B - Adjudication

Read the ledger off your restoration and find the mis-transcribed digit.

Produce two things:

`digits`, shape `(512,)`, integer, your best per-crop digit in ledger order.

`ledger`, a string `"R<row>:<position>:<digit>"` - the failing row, the position in `d0`..`d6`
that was mis-transcribed, and what that digit should have been. Row zero-padded to two
characters. Format example: `"R07:5:3"`, meaning row 7, position 5, the digit should be 3.

Your `digits` and your `ledger` are allowed to disagree. If your reader is uncertain about one
crop but the checksum and the formulary between them pin the row down anyway, submit the pinned
answer. The check digit is not decoration; it is a code with structure, and the structure is
worth more than another point of raw accuracy.

---

## How this is scored

One number. **Lower is better. Zero is perfect. There is no ceiling.**

It is in the NRMSE family and it contains further terms that are not disclosed. What is
disclosed:

- it is computed against `x0`, which you do not have;
- the all-zero restoration is the reference point and scores **1.0** on the restoration term,
  so a restoration term above 1.0 means you have added error rather than removed it;
- Track B enters the score both multiplicatively and additively;
- it is not the average you can compute on the calibration strip, and it does not reward the
  same things. Optimising the strip metric hard will cost you.

You are not told the weighting, the pooling, or the tail handling. That is the point. A metric
you can see is a metric you will fit instead of the problem.

---
## The calibration strip

Twenty-four paired crops. It is the only supervision you have, so use it - but read the
warning printed underneath it before you trust the number it gives you.

````python
def probe(restore_fn, verbose=True):
    """Plain global NRMSE of restore_fn over the calibration strip.

    This is NOT the leaderboard metric. It is a different, simpler, more forgiving
    statistic on a different, easier sample. Treat it as a smoke test."""
    Yh = np.clip(np.stack([restore_fn(z) for z in CALIB_NOISY]), 0, 1)
    num = np.sqrt(np.mean((Yh - CALIB_CLEAN) ** 2))
    den = np.sqrt(np.mean(CALIB_CLEAN ** 2)) + 1e-8
    v = float(num / den)
    if verbose:
        print("probe NRMSE %.4f   (all-zero baseline %.4f)" % (v, 1.0))
        print("  the strip was imaged at a lower noise level than the archive, and this")
        print("  statistic pools differently from the leaderboard. It is an upper bound on")
        print("  how good you are, never an estimate of your score.")
    return v


probe(lambda z: np.clip(z, 0, 1))          # the raw archive, as a floor to beat
````

## My approach (read before running)

Two tracks, one strategy each:

**Track A (restoration).** The chain is blind (`t` varies per crop, params never
recorded) and not invertible, so I don't try to invert it analytically. Instead I
build a *forward simulator* from the given `warp`/`hblur` + the noise schedule,
run it over `X_TRAIN`/`X_DIG` to generate as many `(noisy, clean)` pairs as I want,
and train a regressor to invert that distribution. Because `t` is unknown at
inference time, the model has to be *blind* — trained across the full range of `t`
so it implicitly infers the noise level from the input statistics, the same way a
real blind denoiser would.

**Track B (adjudication).** I don't trust the classifier's raw accuracy — with
per-digit accuracy well under 100%, most of the 64 rows will *look* broken purely
from misreads (this is exactly the trap in Claim 4 in the appendix: at 70% you'd
expect almost every row to have at least one wrong digit by chance). So instead of
"find rows that fail checksum," I do a *structured decode*: use the two free digits
(`d1,d2` = the bed number, which I know exactly from the row index — no reading
required), search for the minimal single-digit edit that makes each row legal +
checksum-consistent, and only then look at *how surprising* that edit was to the
classifier. A row that needed a real correction (the true anomaly) will need to
overrule a *confident, wrong* classifier read — because the archive photographed
the mistake itself, faithfully. A row that only needed fixing because my reader
flubbed a stroke will typically have the true digit sitting as a plausible
runner-up in the probability distribution, so the "cost" of that fix is small.
High-cost + uniquely-determined repairs are my prime suspects.

I don't have ground truth for any of this, so treat every constant below (noise
ranges, model sizes, sample counts) as a knob to tune against `probe()` and your
own judgment — not as a solved problem. I call out the guesses inline.

### Step 1 — forward simulator (fills in what `saltpepper` must be doing)

The chain names `saltpepper(z, p)` but never defines it in the GIVEN cell — only `warp`/`hblur` are handed to us verbatim. I implement a standard salt-and-pepper: flip a `p`-fraction of pixels to the stored range's extremes (`-0.35` / `1.35`). This is a guess; if your calibration (Step 2) suggests otherwise, change it here.

````python
import numpy as np

# ---- guessed parameter ranges — CALIBRATE THESE against the strip in Step 2 ----
RANGE = dict(
    theta=(-15, 15),      # degrees
    scale=(0.85, 1.15),
    dx=(-2.0, 2.0),
    dy=(-2.0, 2.0),
    L_choices=(1, 3, 5, 7),
    g=(0.7, 1.3),
    b=(-0.05, 0.05),
    t=(0, T_STEPS),        # cosine-schedule step; full range, see note below
    p=(0.0, 0.05),          # salt-pepper fraction
)
# NOTE: widened t to the full schedule (0..T_STEPS) — the archive is described
# as noisy "unevenly", and the earlier 0-160 cap silently excluded the hardest
# crops from training entirely, which is a bad way to lose points.

def saltpepper(z, p, rng):
    """Undocumented in GIVEN — my assumption: p-fraction of pixels flipped to
    the stored-range extremes."""
    if p <= 0:
        return z
    out = z.copy()
    mask = rng.random(out.shape) < p
    lo_hi = rng.random(out.shape) < 0.5
    out[mask & lo_hi] = -0.35
    out[mask & ~lo_hi] = 1.35
    return out

def sample_params(rng):
    R = RANGE
    return dict(
        theta=rng.uniform(*R["theta"]),
        scale=rng.uniform(*R["scale"]),
        dx=rng.uniform(*R["dx"]),
        dy=rng.uniform(*R["dy"]),
        L=int(rng.choice(R["L_choices"])),
        g=rng.uniform(*R["g"]),
        b=rng.uniform(*R["b"]),
        t=int(rng.integers(*R["t"])),
        p=rng.uniform(*R["p"]),
    )

def forward_chain(s, params, rng):
    """Exactly the chain in the markdown cell, using the GIVEN warp/hblur."""
    x0 = np.clip(warp(s, params["theta"], params["scale"], params["dx"], params["dy"]), 0, 1)
    u = params["g"] * hblur(x0, params["L"]) + params["b"]
    abar = cosine_alpha_bar(params["t"])
    eps = rng.standard_normal(s.shape)
    z = np.sqrt(abar) * u + np.sqrt(1 - abar) * eps
    z = saltpepper(z, params["p"], rng)
    z = np.clip(z, -0.35, 1.35)
    return x0, z

print("forward_chain ready")
````

### Step 2 — calibrate against the strip (with the paper's warning in mind)

The strip is imaged at a *lower* noise level than the archive and pools differently, so I only use it to sanity-check that my simulated noise is roughly the same *flavor* (similar dynamic range, similar streaking) — not to tune `t` tightly, since the archive almost certainly reaches higher `t` than the strip ever shows.

````python
rng = np.random.default_rng(0)

def quick_synth_batch(n, param_range=None):
    idx = rng.integers(0, X_TRAIN.shape[0], size=n)
    srcs = X_TRAIN[idx]
    X0, Z = [], []
    for s in srcs:
        params = sample_params(rng)
        x0, z = forward_chain(s, params, rng)
        X0.append(x0); Z.append(z)
    return np.stack(X0), np.stack(Z)

_, Zsyn = quick_synth_batch(200)
print("synthetic z   mean %.3f std %.3f  min %.3f max %.3f" %
      (Zsyn.mean(), Zsyn.std(), Zsyn.min(), Zsyn.max()))
print("strip  z      mean %.3f std %.3f  min %.3f max %.3f" %
      (CALIB_NOISY.mean(), CALIB_NOISY.std(), CALIB_NOISY.min(), CALIB_NOISY.max()))
print("archive Z     mean %.3f std %.3f  min %.3f max %.3f" %
      (Z.mean(), Z.std(), Z.min(), Z.max()))
# If archive Z is visibly noisier/streakier than strip (it should be, per the
# markdown), that's expected. If your synthetic Zsyn doesn't bracket both,
# widen RANGE above and re-run this cell.

def synth_nrmse_eval(restore_fn, n=800, seed=999):
    """NRMSE against x0 on a FRESH synthetic batch spanning the full t range —
    unlike probe(), this actually exercises high-noise crops, which the strip
    (per the markdown) does not. Use this, not probe() alone, to judge whether
    a change actually helped."""
    r = np.random.default_rng(seed)
    idx = r.integers(0, X_TRAIN.shape[0], size=n)
    X0v, Zv = [], []
    for s in X_TRAIN[idx]:
        params = sample_params(r)
        x0, z = forward_chain(s, params, r)
        X0v.append(x0); Zv.append(z)
    X0v, Zv = np.stack(X0v), np.stack(Zv)
    Yh = np.clip(np.stack([restore_fn(z) for z in Zv]), 0, 1)
    num = np.sqrt(np.mean((Yh - X0v) ** 2))
    den = np.sqrt(np.mean(X0v ** 2)) + 1e-8
    nrmse = float(num / den)
    baseline = np.sqrt(np.mean(Zv ** 2)) / den   # roughly what all-zero-vs-scale looks like
    print("held-out synthetic NRMSE: %.4f   (identity/no-op baseline ~%.4f)" % (nrmse, baseline))
    print("  remember: the leaderboard's all-ZERO restoration scores exactly 1.0 on this")
    print("  term, so getting meaningfully under that number is the actual bar to clear.")
    return nrmse
````

### Step 3 — restoration model (Track A), now with a real CNN

Kaggle notebooks ship PyTorch (and usually GPU) preinstalled, so there's no reason
to stay on a flattened-pixel MLP — a small conv autoencoder respects the 2D
structure of the strokes and should denoise substantially better for the same
training budget. I train a small **ensemble of 2** independently-seeded CNNs and
average their output at inference time, which reduces variance for free.

Important framing from the scoring cell: **the all-zero restoration scores exactly
1.0** on the restoration term. If your NRMSE isn't clearly under that, you're not
beating the "do nothing" baseline. I validate against a fresh held-out synthetic
batch (`synth_nrmse_eval`, Step 2) rather than trusting `probe()` alone, since the
markdown explicitly warns the strip doesn't reward the same things as the real
score.

````python
import torch, torch.nn as nn, torch.nn.functional as F

device = "cuda" if torch.cuda.is_available() else "cpu"
print("device:", device)

class DenoiserCNN(nn.Module):
    """Small conv autoencoder, 28x28 -> 14x14 -> 7x7 -> 14x14 -> 28x28, with one
    residual skip from the first encoder layer into the last decoder layer.
    BatchNorm + LeakyReLU throughout: plain ReLU with no normalization is prone
    to "dying" (activations collapse to all-negative, gradients go to zero,
    loss plateaus at a constant-output solution) — which is what the flat MSE
    plateau you saw was."""
    def __init__(self):
        super().__init__()
        # widened vs. the first version (32/64/128 -> 48/96/192) — more capacity
        # to close the gap between held-out NRMSE and what the archive needs
        self.enc1 = nn.Conv2d(1, 48, 3, padding=1);  self.bn1 = nn.BatchNorm2d(48)
        self.enc2 = nn.Conv2d(48, 96, 3, padding=1, stride=2);  self.bn2 = nn.BatchNorm2d(96)   # -> 14x14
        self.enc3 = nn.Conv2d(96, 192, 3, padding=1, stride=2); self.bn3 = nn.BatchNorm2d(192)  # -> 7x7
        self.dec3 = nn.ConvTranspose2d(192, 96, 3, stride=2, padding=1, output_padding=1); self.bn4 = nn.BatchNorm2d(96)
        self.dec2 = nn.ConvTranspose2d(96, 48, 3, stride=2, padding=1, output_padding=1);  self.bn5 = nn.BatchNorm2d(48)
        self.dec1 = nn.Conv2d(48, 1, 3, padding=1)
        self.act = nn.LeakyReLU(0.1)

    def forward(self, x):
        h1 = self.act(self.bn1(self.enc1(x)))
        h2 = self.act(self.bn2(self.enc2(h1)))
        h3 = self.act(self.bn3(self.enc3(h2)))
        d3 = self.act(self.bn4(self.dec3(h3)))
        d2 = self.act(self.bn5(self.dec2(d3)) + h1)     # residual skip
        return self.dec1(d2)   # linear output — no sigmoid. Sigmoid + MSE on a
        # continuous [0,1] target is the wrong pairing: if the pre-activation
        # drifts very negative early (easy with random init), sigmoid's
        # gradient vanishes there and the layer can never recover, no matter
        # how healthy the batchnorm'd layers before it are. We already clip
        # to [0,1] at inference time (restore_fn / restore_batch), so a plain
        # linear output plus that clip is the safer combination.

import gc

def train_denoiser(n_samples=70000, epochs=32, batch_size=256):
    X0n, Zn = quick_synth_batch(n_samples)
    X0t = torch.tensor(X0n, dtype=torch.float32).unsqueeze(1)
    Zt  = torch.tensor(Zn,  dtype=torch.float32).unsqueeze(1)
    model = DenoiserCNN().to(device)
    opt = torch.optim.Adam(model.parameters(), lr=1e-3)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    n = X0t.shape[0]
    for ep in range(epochs):
        perm = torch.randperm(n)
        tot = 0.0
        for i in range(0, n, batch_size):
            idx = perm[i:i+batch_size]
            xb, yb = Zt[idx].to(device), X0t[idx].to(device)
            opt.zero_grad()
            loss = F.mse_loss(model(xb), yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            opt.step()
            tot += loss.item() * xb.size(0)
            del xb, yb
        sched.step()
        if ep % 4 == 0 or ep == epochs - 1:
            print(f"  epoch {ep:2d}  train mse {tot/n:.5f}  lr {sched.get_last_lr()[0]:.2e}")
    model.eval()
    del X0t, Zt, opt
    gc.collect()
    if device == "cuda":
        torch.cuda.empty_cache()
    return model

def finetune_on_strip(model, epochs=40, lr=1e-4):
    """The 24 calibration pairs are REAL ground truth — not a guess about
    corruption parameters like everything else. But it's a tiny dataset at a
    (per the markdown) different noise level than the archive, so this is a
    gentle nudge with a low LR and no BatchNorm-destabilizing large steps —
    not something to lean on hard. Skip this if it makes synth_nrmse_eval
    WORSE — that would mean the strip's distribution is different enough from
    the archive that fitting it is actively hurting you, exactly the trap the
    markdown warns about."""
    X0t = torch.tensor(CALIB_CLEAN, dtype=torch.float32).unsqueeze(1).to(device)
    Zt  = torch.tensor(CALIB_NOISY, dtype=torch.float32).unsqueeze(1).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    model.train()
    for ep in range(epochs):
        opt.zero_grad()
        loss = F.mse_loss(model(Zt), X0t)
        loss.backward()
        opt.step()
    model.eval()
    return model

print("training denoiser ensemble member 1/3 ...")
d_model_a = train_denoiser()
print("training denoiser ensemble member 2/3 ...")
d_model_b = train_denoiser()
print("training denoiser ensemble member 3/3 ...")
d_model_c = train_denoiser()
denoiser_models = [d_model_a, d_model_b, d_model_c]

@torch.no_grad()
def restore_fn(z):
    zt = torch.tensor(z, dtype=torch.float32).reshape(1, 1, 28, 28).to(device)
    preds = [m(zt).cpu().numpy().reshape(28, 28) for m in denoiser_models]
    return np.clip(np.mean(preds, axis=0), 0, 1)

print("--- before strip fine-tune ---")
probe(restore_fn)
nrmse_before = synth_nrmse_eval(restore_fn)

# fine-tune each ensemble member on the 24 real pairs, then keep it ONLY if it
# actually helps the held-out synthetic score — snapshot weights first since
# fine-tuning mutates the models in place and we need a real way to revert
import copy
snapshots = [copy.deepcopy(m.state_dict()) for m in denoiser_models]
for m in denoiser_models:
    finetune_on_strip(m)
print("--- after strip fine-tune ---")
probe(restore_fn)
nrmse_after = synth_nrmse_eval(restore_fn)

if nrmse_after > nrmse_before:
    print(f"strip fine-tune made held-out NRMSE worse ({nrmse_before:.4f} -> {nrmse_after:.4f})"
          " — reverting to the pre-fine-tune weights.")
    for m, snap in zip(denoiser_models, snapshots):
        m.load_state_dict(snap)
        m.eval()
else:
    print(f"strip fine-tune helped ({nrmse_before:.4f} -> {nrmse_after:.4f}) — keeping it.")
````

````python
@torch.no_grad()
def restore_batch(Zb, chunk=1024):
    """Restore in chunks — pushing tens of thousands of crops through the
    ensemble in one shot is what caused the CUDA OOM. This also frees each
    chunk's activations immediately instead of holding everything at once."""
    outs = []
    for i in range(0, len(Zb), chunk):
        zt = torch.tensor(Zb[i:i+chunk], dtype=torch.float32).unsqueeze(1).to(device)
        preds = [m(zt).cpu().numpy()[:, 0] for m in denoiser_models]
        outs.append(np.clip(np.mean(preds, axis=0), 0, 1))
        del zt
    if device == "cuda":
        torch.cuda.empty_cache()
    return np.concatenate(outs, axis=0)

restored = restore_batch(Z).astype(np.float32)
print("restored:", restored.shape, "mean %.4f" % restored.mean())
````

### Step 4 — digit classifier: CNN, trained on restored crops, then fine-tuned on REAL labels

Same architecture upgrade as Track A: a small conv classifier instead of a flat
MLP, still trained on synthetic crops pushed through the trained denoiser so it
learns the denoiser's actual artifact distribution.

Then a second pass that isn't in the original plan: **the bed-number digits
(`d1`, `d2` of every row) are known exactly from the row index — no model
needed.** That's 128 crops of *real* archive data with *certain* ground-truth
labels, for free. I fine-tune the classifier on those after the synthetic
pretraining. It's a small dataset so I use a low learning rate and few epochs to
nudge the model toward the archive's real noise/artifact statistics without
overfitting 128 examples — but even a small nudge in the right direction should
help every other crop's reading, since it's genuine domain signal instead of a
guess about corruption parameters.

````python
class ClassifierCNN(nn.Module):
    def __init__(self):
        super().__init__()
        # widened (32/64 -> 64/128) plus an extra FC layer — the single-model
        # version was still improving at epoch 35, so it was capacity/time
        # limited, not converged
        self.c1 = nn.Conv2d(1, 64, 3, padding=1);   self.bn1 = nn.BatchNorm2d(64)
        self.c2 = nn.Conv2d(64, 128, 3, padding=1); self.bn2 = nn.BatchNorm2d(128)
        self.fc1 = nn.Linear(128 * 7 * 7, 256)
        self.fc2 = nn.Linear(256, 64)
        self.fc3 = nn.Linear(64, 10)
        self.drop = nn.Dropout(0.3)
        self.act = nn.LeakyReLU(0.1)

    def forward(self, x):
        x = F.max_pool2d(self.act(self.bn1(self.c1(x))), 2)   # -> 14x14
        x = F.max_pool2d(self.act(self.bn2(self.c2(x))), 2)   # -> 7x7
        x = x.flatten(1)
        x = self.drop(self.act(self.fc1(x)))
        x = self.drop(self.act(self.fc2(x)))
        return self.fc3(x)

def labeled_synth_batch(n):
    pool_X = np.concatenate([X_TRAIN, X_DIG], axis=0)
    pool_y = np.concatenate([Y_TRAIN, Y_DIG], axis=0)
    idx = rng.integers(0, pool_X.shape[0], size=n)
    srcs, labels = pool_X[idx], pool_y[idx]
    Zb = np.stack([forward_chain(s, sample_params(rng), rng)[1] for s in srcs])
    return Zb, labels

def run_epochs(model, opt, X_t, y_t, epochs, batch_size=256, lr=None, tag="", sched=None):
    if lr is not None:
        for g in opt.param_groups:
            g["lr"] = lr
    n = X_t.shape[0]
    for ep in range(epochs):
        perm = torch.randperm(n)
        tot = correct = 0
        for i in range(0, n, batch_size):
            idx = perm[i:i+batch_size]
            xb, yb = X_t[idx].to(device), y_t[idx].to(device)
            opt.zero_grad()
            out = model(xb)
            loss = F.cross_entropy(out, yb)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
            opt.step()
            tot += loss.item() * xb.size(0)
            correct += (out.argmax(1) == yb).sum().item()
        if sched is not None:
            sched.step()
        if ep % 3 == 0 or ep == epochs - 1:
            print(f"{tag} epoch {ep:2d}  loss {tot/n:.4f}  acc {correct/n:.4f}")

# ---- free real supervision from the known bed-number digits (used by every
# ensemble member below) ----
bed_idx, bed_labels = [], []
for r in range(64):
    d1, d2 = divmod(r + 1, 10)
    bed_idx += [8*r + 1, 8*r + 2]
    bed_labels += [d1, d2]
bed_labels_arr = np.array(bed_labels)

def train_one_classifier(n_samples=60000, epochs=50, ft_epochs=25):
    Zc, yc = labeled_synth_batch(n_samples)
    Rc = restore_batch(Zc)     # restore with the trained denoiser BEFORE training the reader
    clf = ClassifierCNN().to(device)
    opt = torch.optim.Adam(clf.parameters(), lr=1e-3)
    sched = torch.optim.lr_scheduler.CosineAnnealingLR(opt, T_max=epochs)
    Xc_t = torch.tensor(Rc, dtype=torch.float32).unsqueeze(1)
    yc_t = torch.tensor(yc, dtype=torch.long)
    run_epochs(clf, opt, Xc_t, yc_t, epochs=epochs, tag="[synthetic]", sched=sched)

    Xb_t = torch.tensor(restored[bed_idx], dtype=torch.float32).unsqueeze(1)
    yb_t = torch.tensor(bed_labels, dtype=torch.long)
    run_epochs(clf, opt, Xb_t, yb_t, epochs=ft_epochs, batch_size=32, lr=2e-4,
               tag="[real bed-digit fine-tune]")
    clf.eval()
    del Xc_t, yc_t, Rc
    gc.collect()
    if device == "cuda":
        torch.cuda.empty_cache()
    return clf

@torch.no_grad()
def synth_clf_accuracy(models, n=3000, seed=777):
    """Held-out generalization check — NOT the bed digits (those are the
    fine-tune set, so of course they're near 100%). This is what actually
    predicts how good your reading is on the other 384 crops."""
    r = np.random.default_rng(seed)
    pool_X = np.concatenate([X_TRAIN, X_DIG], axis=0)
    pool_y = np.concatenate([Y_TRAIN, Y_DIG], axis=0)
    idx = r.integers(0, pool_X.shape[0], size=n)
    srcs, labels = pool_X[idx], pool_y[idx]
    Zh = np.stack([forward_chain(s, sample_params(r), r)[1] for s in srcs])
    Rh = restore_batch(Zh)
    Rh_t = torch.tensor(Rh, dtype=torch.float32).unsqueeze(1).to(device)
    all_p = [F.softmax(m(Rh_t), dim=1).cpu().numpy() for m in models]
    pred = np.mean(all_p, axis=0).argmax(axis=1)
    acc = (pred == labels).mean()
    print("held-out synthetic classifier accuracy (ensemble): %.4f" % acc)
    return acc

# a SINGLE classifier's Track-B ranking moved a lot between runs last time —
# that instability means don't trust any one run. Average several independently
# -trained classifiers instead, same idea as the denoiser ensemble in Step 3.
N_CLF_MODELS = 3
clf_models = []
for i in range(N_CLF_MODELS):
    print(f"training classifier {i+1}/{N_CLF_MODELS} ...")
    clf_models.append(train_one_classifier())

with torch.no_grad():
    R512_t = torch.tensor(restored, dtype=torch.float32).unsqueeze(1).to(device)
    all_probs = [F.softmax(m(R512_t), dim=1).cpu().numpy() for m in clf_models]
probs = np.mean(all_probs, axis=0)          # averaged ensemble probabilities
argmax_digits = probs.argmax(axis=1)
print("probs:", probs.shape)
print("ensemble accuracy on the 128 KNOWN bed digits (fine-tune set, expect ~1.0):",
      (argmax_digits[bed_idx] == bed_labels_arr).mean())
per_model_acc = [( (p.argmax(axis=1)[bed_idx] == bed_labels_arr).mean() ) for p in all_probs]
print("  (individual models on bed digits:", [round(a, 3) for a in per_model_acc], ")")
synth_clf_accuracy(clf_models)   # <- this is the number that actually matters
````

### Step 5 — Track B: structured ledger decode

For every row: hard-set the two bed-number digits (free, no reading needed), take
the classifier's argmax elsewhere, and if that's not legal+checksum-consistent,
search all single-digit edits at the remaining positions for the minimal-cost fix.
Then rank candidate anomaly rows by (a) how *surprising* the fix was to the
classifier — high cost means the model was confidently wrong, which only happens
when the image itself shows the mis-transcribed digit — and (b) whether the legal
repair is *unique*, since a row with several equally-legal one-digit fixes is more
likely just reader noise than a real anomaly.

````python
import itertools, math

EPS = 1e-9
log_probs = np.log(probs + EPS)     # (512, 10)

digits = argmax_digits.copy()
row_reports = []

for r in range(64):
    d1_true, d2_true = divmod(r + 1, 10)
    base = list(argmax_digits[8*r:8*r+8])
    base[1], base[2] = d1_true, d2_true   # free digits, always correct
    digits[8*r+1] = d1_true
    digits[8*r+2] = d2_true

    def legal_and_consistent(d8):
        return row_is_legal(r, d8) and check_digit(d8[:7]) == d8[7]

    if legal_and_consistent(base):
        row_reports.append(dict(row=r, needs_fix=False))
        continue

    # search single-digit edits at the 6 unknown positions (0,3,4,5,6,7)
    candidates = []
    for pos in (0, 3, 4, 5, 6, 7):
        for cand_digit in range(10):
            trial = base.copy()
            trial[pos] = cand_digit
            if legal_and_consistent(trial):
                cost = -log_probs[8*r + pos, cand_digit]
                candidates.append((cost, pos, cand_digit))

    if not candidates:
        row_reports.append(dict(row=r, needs_fix=True, resolved=False))
        continue

    candidates.sort()
    best_cost, best_pos, best_digit = candidates[0]
    n_equally_legal = sum(1 for c in candidates if c[0] <= best_cost + 1e-6)
    row_reports.append(dict(row=r, needs_fix=True, resolved=True,
                             pos=best_pos, digit=best_digit,
                             cost=best_cost, n_options=len(candidates),
                             unique=(n_equally_legal == 1)))
    digits[8*r + best_pos] = best_digit   # snap to the legal repair for the `digits` output

# rank resolved, unique repairs by cost (descending) — top of this list is
# the strongest candidate for the *real* anomaly
ranked = sorted(
    [rr for rr in row_reports if rr.get("resolved") and rr["unique"]],
    key=lambda rr: -rr["cost"],
)
print(f"{sum(r['needs_fix'] for r in row_reports)} / 64 rows needed a fix")
print(f"{len(ranked)} rows had a UNIQUE single-digit legal repair")
for rr in ranked[:8]:
    print(f"  row {rr['row']:02d}  pos {rr['pos']}  -> digit {rr['digit']}"
          f"   cost(-log p)={rr['cost']:.2f}   n_legal_options={rr['n_options']}")
````

````python
# Inspect the top few candidates yourself before trusting the top one blindly —
# print the actual restored crop at (row, pos) alongside the classifier's full
# distribution, and sanity check against the formulary/dose range by eye.
import matplotlib.pyplot as plt

top = ranked[0]
r, pos = top["row"], top["pos"]
crop_idx = 8*r + pos
fig, ax = plt.subplots(1, 2, figsize=(6, 3))
ax[0].imshow(restored[crop_idx], cmap="gray", vmin=0, vmax=1)
ax[0].set_title(f"row {r} pos {pos}\nrepair -> {top['digit']}")
ax[0].axis("off")
ax[1].bar(range(10), probs[crop_idx])
ax[1].set_xticks(range(10)); ax[1].set_title("classifier probs")
plt.tight_layout(); plt.show()

row_str = f"R{top['row']:02d}:{top['pos']}:{top['digit']}"
print("ledger candidate:", row_str)
````

````python
ledger = row_str   # promote once you've eyeballed it against the plot above
print("digits:", digits.shape, np.bincount(digits, minlength=10))
print("ledger:", ledger)
````

## Submitting

Run the cell below once you have all three objects. It validates first and refuses to write a
submission that would be rejected on the leaderboard, so a clean run here means a clean upload.

It writes `submission.csv` - 401,923 rows, about 7 MB. Upload that file on the competition's
**Submit Predictions** page.

Keep this notebook with your working in it and all outputs present. A submission whose notebook
does not reproduce it is not a submission.

````python
# ==================== SUBMISSION - run last, writes submission.csv ====================
def write_submission(restored, digits, ledger, path="submission.csv"):
    """Writes the Kaggle submission. Validates first, so a clean run here is a
    clean upload. Expects the same three objects as before."""
    import re
    R = np.asarray(restored, dtype=np.float32)
    D = np.asarray(digits).astype(np.int64).reshape(-1)
    L = str(ledger).strip().upper()

    assert R.shape == (512, 28, 28), f"restored must be (512,28,28), got {R.shape}"
    assert np.isfinite(R).all(),      "restored contains NaN or inf"
    assert D.shape == (512,),         f"digits must be (512,), got {D.shape}"
    assert D.min() >= 0 and D.max() <= 9, "digits must lie in 0..9"
    m = re.fullmatch(r"R(\d{2}):([0-6]):(\d)", L)
    assert m, 'ledger must look like "R07:5:3"'
    row, pos, dig = int(m.group(1)), int(m.group(2)), int(m.group(3))
    assert 0 <= row <= 63, "ledger row out of range"

    if R.min() < 0 or R.max() > 1:
        print("warning: restored outside [0,1] (%.3f, %.3f); it will be clipped"
              % (R.min(), R.max()))
    R = np.clip(R, 0, 1)

    with open(path, "w") as f:
        f.write("Id,Value\n")
        for c in range(512):
            flat = R[c].ravel()
            for p in range(784):
                f.write(f"P_{c:03d}_{p:03d},{flat[p]:.4f}\n")
        for c in range(512):
            f.write(f"D_{c:03d},{int(D[c])}\n")
        f.write(f"L_ROW,{row}\nL_POS,{pos}\nL_DIG,{dig}\n")
        f.write(f"LP_ROW,{row}\nLP_POS,{pos}\nLP_DIG,{dig}\n")

    n = 512 * 784 + 512 + 6
    print(f"wrote {path}  ({n} rows)")
    print("  restored  mean %.4f" % R.mean())
    print("  digits    ", np.bincount(D, minlength=10))
    print("  ledger    ", L)
    print("\nUpload this file on the competition's Submit Predictions page.")



write_submission(restored, digits, ledger)
````

## Appendix - what was already concluded

Read this after you have your own numbers, not before.

Two teams looked at this archive before it reached you. Between them they make six numbered
claims. **Exactly one of the six is true.** Every one of them is decidable with code you will
have written by then, and one of them is worth real money to you.

---

**Imaging & Archive, 3 July**

> We characterised the recovered tensor against the surviving calibration strip.
>
> **(1)** The archive is homogeneous. A denoiser fitted to the calibration strip transfers to
> the scored crops without adjustment.
> **(2)** Nothing was lost to clipping. The stored range is the natural dynamic range of the
> sensor, not a saturation artefact.
> **(3)** The crops were written by the same cohort of scribes as the public training split,
> so a reader validated on that split is validated here.

---

**Ledger Assurance, 5 July**

> We modelled the identification problem before committing reader effort.
>
> **(4)** The check digit localises the failing row on its own. A reader at seventy percent
> per-digit accuracy is sufficient to find it.
> **(5)** The failing row can be found without reading a single digit, by ranking rows on
> recovered image quality - a mis-transcription leaves a trace in the scan.
> **(6)** Under the published constraints - ward set, bed anchor, formulary, dose range - the
> failing row admits exactly one legal single-digit repair.

---

Claim (4) is the one that cost the most. It is arithmetically reasonable and it is wrong for a
reason that has nothing to do with arithmetic: work out how many of the sixty-four rows survive
a reader at that accuracy without a single error in them, and you will see what the assurance
team saw on their screen when they ran it.

---

## A closing note on the shape of this problem

The Kannada Ward was an inverse problem in one variable at a time: what `k`, what width, what
position. This one is an inverse problem in a chain. Nine unknowns per crop, five hundred and
twelve crops, one paired example set of twenty-four, and a metric you cannot see.

You cannot solve it by inverting the chain, because the chain is not invertible. You solve it
by building a forward model you *can* run - you were handed `warp`, `hblur` and the schedule
for exactly this reason - running it against the corpus until you have as much matched data as
you want, and learning the inverse from that. The blindness is the problem; the fact that `t`
varies per crop is the problem; the clip is the floor.

And then, having done all that, you will still be at eighty-something percent on the digits,
and you will find the failing row anyway - not because your reader got that row right, but
because sixty-three other rows close and one does not, and a code with structure will tell you
things your classifier never could.

That is the part worth learning. The archive is noisy. The ledger is not.
